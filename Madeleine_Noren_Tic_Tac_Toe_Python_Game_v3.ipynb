```python
## Pre-requisites
from IPython.display import clear_output  # Refreshes the screen at every stage so that instructions, each input, and final stats are not output as a long list
```

```python
## Game Setup - Global Counters

games_played = 0    # number of finished games
x_wins = 0          # number of games player X won
o_wins = 0          # number of games player O won
draws = 0           # number of games drawn
total_moves = 0     # total number of moves across all games / maybe in each game?

# All 8 combinations of winning a game - read left to right, top to bottom (0 = top left, 8 = bottom right)
winning_combinations = [
    [0,1,2], [3,4,5], [6,7,8],    # rows
    [0,3,6], [1,4,7], [2,5,8],    # columns
    [0,4,8], [2,4,6]    # diagonals
]
```

```python
## Instructions and "Are you ready?"

def show_instructions():
    'Print the rules and explain how to choose a square.'
    print('=================================')
    print('     Welcome to Tic Tac Toe!')
    print('=================================')
    print('Two players take turns: X goes first, then O.')
    print('Pick a square by typing its number (1-9):')
    print()
    print(' 1 | 2 | 3 ')
    print('---+---+---')
    print(' 4 | 5 | 6 ')
    print('---+---+---')
    print(' 7 | 8 | 9 ')
    print()
    print('Get three in a row (across, down or diagonally) to win!')
    print('Type q at any time to quit.')
    print()

# Ask if the players are ready to start - return true for yes
def ask_if_ready():
  while True:
    choice = input('Are you ready to play? (y/n): ').strip().lower()
    if choice == 'y':
      return True
    elif choice == 'n' or choice == 'q':  # Option to quit as well
      return False
    print('Please type y or n.')
```

```python
## Game board

# Create a new board where players can see which square 1-9 to pick/type
# Will make the numbers grey (as to not distract), Xs orange and Os blue (for colour-blind people)
def create_board():
  return ['1','2','3','4','5','6','7','8','9']

# Print the board as a 3x3 grid which holds the numbers 1-9
def display_board(board):
  print()
  print(f' {board[0]} | {board[1]} | {board[2]} ')
  print('---+---+---')
  print(f' {board[3]} | {board[4]} | {board[5]} ')
  print('---+---+---')
  print(f' {board[6]} | {board[7]} | {board[8]} ')
  print()
```

```python
## Getting a valid move from the player

def get_move(board, player):
  valid_move = False    # Becomes true when the player picks a valid move
  position = 0    # Will hold the chosen position in the board list
  while valid_move == False:
    choice = input('Player ' + player + ', enter your move (1-9): ').strip()    # .stip() removes any whitespaces in input
    if choice == 'q':   # Option to quit
      return -1     # Not a real position, so it gets handled as a quit in the play_game function
    if choice.isdigit() == False:    # Check that input is a number
      error_message = 'That is not a number. Please enter a number from 1 to 9.'
    elif int(choice) < 1 or int(choice) > 9:    # Check that input is from 1 to 9
      error_message = 'That number is out of range. Please enter a number from 1 to 9.'
    elif board[int(choice) - 1] == 'X' or board[int(choice) - 1] == 'O':  # Check that the square chosen is free
      error_message = 'That square is already taken. Please choose another one.'
    else:
      position = int(choice) - 1    # -1 because the actual positions start 0-8, not 1-9
      valid_move = True   # Ends the while loop
    if valid_move == False:   # Clear screen and show board again if move not valid
      clear_output()    # Refreshes screen
      display_board(board)
      print(error_message)    # Outputs specific error message
  return position
```

```python
## Check for 3 in a row - True if won, False otherwise

def check_winner(board, player):
  for combination in winning_combinations:
    if board[combination[0]] == board[combination[1]] == board[combination[2]] == player:
      return True
  return False
```

```python
## Check if the board is full e.g. leading to a draw

def is_board_full(board):
  for square in board:
    if square != 'X' and square != 'O':
      return False
  return True
```

```python
## Changing players

def switch_player(player):
  if player == 'X':
    return 'O'
  else:
    return 'X'
```

```python
## Update global counters when a game finishes

def update_statistics(result):
  global games_played, x_wins, o_wins, draws
  games_played += 1
  if result == 'X':
    x_wins += 1
  elif result == 'O':
    o_wins += 1
  else:
    draws += 1
```

```python
## Print the statistics for all games played so far

def show_statistics():
  print('=========================')
  print('     Game Statistics')
  print('=========================')
  print(f'Games played: {games_played}')
  print(f'Player X wins: {x_wins}')
  print(f'Player O wins: {o_wins}')
  print(f'Draws: {draws}')
  print(f'Total moves: {total_moves}')
  print()
```

```python
## Playing a game and returning the result

def play_game():
  global total_moves
  board = create_board()
  current_player = 'X'    # Player X always starts first
  while True:
    clear_output()    # Refresh screen
    display_board(board)
    position = get_move(board, current_player)
    if position == -1:    # Player has quit
      return 'quit'
    board[position] = current_player
    total_moves += 1
    if check_winner(board, current_player):   # check for a winner after each move
      clear_output()    # Refresh screen
      display_board(board)
      print('Player ' + current_player + ' wins!')
      return current_player
    if is_board_full(board):    # also check if the board is full after each move
      clear_output()    # Refresh screen
      display_board(board)
      print("It's a draw!")   # double quotes because of the apostrophe
      return 'draw'
    current_player = switch_player(current_player)
```

```python
## Ask if the players want to play again - return true for yes

def ask_to_play_again():
  while True:
    choice = input('Do you want to play again? (y/n): ').strip().lower()
    if choice == 'y':
      return True
    elif choice == 'n' or choice == 'q':  # Includes quit option
      return False
    print('Please type y for yes or n for no.')
```

```python
## Run the game

def main():
  show_instructions()
  playing = ask_if_ready()

  while playing:
    result = play_game()
    if result == 'quit':    # Quit option does not count the current game
      playing = False
    else:
      update_statistics(result)
      show_statistics()
      playing = ask_to_play_again()

  clear_output()    #Refresh screen
  print('Thanks for playing!')
  show_statistics()
```

```python
main()
```
